Calcular el sueldo de un empleado. hrs * valor de horas * 30 dias
deducciones: ISSS, AFP, ISR.

```python
calcularSueldo = lambda h, v: float(h) * float(v) * 30
```

```python
horas = input("Horas trabajadas por dia: ")
valor = input("Valor por hora:$")
bruto = calcularSueldo(horas,valor)

isss = bruto * 0.03
afp = bruto * 0.0725
isr = bruto * 0.10

print("Salario Bruto:$ ", bruto)
print("ISSS:$",isss)
print("AFP:$",afp)
print("ISR:$",isr)
print("Salario Neto:$", bruto - isss - afp - isr)
```

```python
persona ={
    "codigo":"angel123",
    "nombre":"angel jimenez",
    "materias":[
    {
      "codigo":"611",
      "nombre":"pogra 1",
      "notas":{
          "C1":{"Lab1":8,"Lab2":9,"Parcial":10},
          "C2":{"Lab1":8,"Lab2":9,"Parcial":10},
          "C3":{"Lab1":8,"Lab2":9,"Parcial":10}
      }
    },
    {
      "codigo":"612",
      "nombre":"pogra ll",
      "notas":{
          "C1":{"Lab1":8,"Lab2":9,"Parcial":10},
          "C2":{"Lab1":8,"Lab2":9,"Parcial":10},
          "C3":{"Lab1":8,"Lab2":9,"Parcial":10}
      }
    }
    ]
}
print(persona["nombre"], "materias: ", persona["materias"][0]["nombre"])

def calcular_promedio_corte(corte_notas):
    """Calcula el promedio de las notas de un corte."""
    total_notas = sum(corte_notas.values())
    cantidad_notas = len(corte_notas)
    return total_notas / cantidad_notas

def calcular_nf_materia(materia):
    """Calcula la nota final (NF) de una materia."""
    notas_cortes = materia["notas"]
    promedios_cortes = []
    for corte in ["C1", "C2", "C3"]:
        if corte in notas_cortes:
            promedios_cortes.append(calcular_promedio_corte(notas_cortes[corte]))

    if promedios_cortes:
        return sum(promedios_cortes) / len(promedios_cortes)
    else:
        return 0 # O manejar el caso sin notas de otra forma

def formatear_nf_materia(materia):
    """Calcula y formatea la Nota Final de una materia."""
    nf = calcular_nf_materia(materia)
    return f"Materia: {materia['nombre']}, Nota Final (NF): {nf:.2f}"

print("\n--- Notas Finales por Materia (usando map) ---")
# Usar map para aplicar la función a cada materia
resultados_nf = map(formatear_nf_materia, persona["materias"])

# Imprimir los resultados
for resultado in resultados_nf:
    print(resultado)
```
