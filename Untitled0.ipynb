Calcular el sueldo de un empleado. hrs * valor de horas * 30 dias
deducciones: ISSS, AFP, ISR.

```python
calcularSueldo = lambda h, v: float(h) * float(v) * 30
```

```python
horas = input("Horas trabajadas por dia: ")
valor = input("Valor por hora:$")
bruto = calcularSueldo(horas,valor)

isss = bruto * 0.03
afp = bruto * 0.0725
isr = bruto * 0.10

print("Salario Bruto:$ ", bruto)
print("ISSS:$",isss)
print("AFP:$",afp)
print("ISR:$",isr)
print("Salario Neto:$", bruto - isss - afp - isr)
```

Elevar un numero a su exponente.
