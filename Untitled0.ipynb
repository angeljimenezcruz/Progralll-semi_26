1- Desarrolle un algoritmo en python que calcule el impuesto a las actividades economicas de acuerdo con las siguiente tabla. el monto de la actividad economica debe ser inducido por el usuario.

```python
def calcular_impuesto():
    print("CALCULADORA DE IMPUESTO A LAS ACTIVIDADES ECONÓMICAS")

    impuestos = [
        {"desde": 0.01, "hasta": 500, "precio": 1.5, "adicional": 0},
        {"desde": 500.01, "hasta": 1000, "precio": 1.5, "adicional": 3},
        {"desde": 1000.01, "hasta": 2000, "precio": 3, "adicional": 3},
        {"desde": 2000.01, "hasta": 3000, "precio": 6, "adicional": 3},
        {"desde": 3000.01, "hasta": 6000, "precio": 9, "adicional": 2},
        {"desde": 6000.01, "hasta": 8000, "precio": 15, "adicional": 2},
        {"desde": 8000.01, "hasta": 18000, "precio": 15, "adicional": 2},
        {"desde": 18000.01, "hasta": 30000, "precio": 39, "adicional": 2},
        {"desde": 30000.01, "hasta": 60000, "precio": 63, "adicional": 1},
        {"desde": 60000.01, "hasta": 100000, "precio": 93, "adicional": 0.8},
        {"desde": 100000.01, "hasta": 200000, "precio": 125, "adicional": 0.7},
        {"desde": 200000.01, "hasta": 300000, "precio": 195, "adicional": 0.6},
        {"desde": 300000.01, "hasta": 400000, "precio": 255, "adicional": 0.45},
        {"desde": 400000.01, "hasta": 500000, "precio": 300, "adicional": 0.4},
        {"desde": 500000.01, "hasta": 1000000, "precio": 340, "adicional": 0.30},
        {"desde": 1000000.01, "hasta": 99999999, "precio": 490, "adicional": 0.18}
    ]

    while True:
        monto = float(input("Ingrese el monto de la actividad económica: $"))

        if monto <= 0:
            print("El monto debe ser mayor que 0.")
        else:
            for fila in impuestos:
                if fila["desde"] <= monto <= fila["hasta"]:
                    impuesto = fila["precio"] + (
                        ((monto - fila["desde"]) / 1000) * fila["adicional"]
                    )
                    print(f"Impuesto a pagar: ${impuesto:.2f}")
                    break
            else:
                print("Monto fuera de rango")

        continuar = input("¿Desea realizar otro cálculo? (s/n): ").lower()
        if continuar != 's':
            break

calcular_impuesto()
```

```python
from functools import reduce
from collections import Counter

class estadistica:

    def media(self, serie):
        suma = reduce(lambda acum, num: acum + float(num), serie, 0)
        return suma / len(serie)

    def mediana(self, serie):
        numeros = sorted([float(num) for num in serie])
        n = len(numeros)
        if n % 2 == 0:
            return (numeros[n//2 - 1] + numeros[n//2]) / 2
        else:
            return numeros[n//2]

    def moda(self, serie):
        numeros = [float(num) for num in serie]
        contador = Counter(numeros)
        max_frec = max(contador.values())
        modas = [k for k, v in contador.items() if v == max_frec]
        return modas[0] if len(modas) == 1 else modas

    def varianza(self, serie):
        media = self.media(serie)
        suma = reduce(lambda acum, num: acum + (float(num) - media) ** 2, serie, 0)
        return suma / len(serie)

    def desviacion(self, serie):
        return self.varianza(serie) ** 0.5

    def rango(self, serie):
        numeros = [float(num) for num in serie]
        return max(numeros) - min(numeros)

    def frecuencia(self, serie):
        numeros = [float(num) for num in serie]
        contador = Counter(numeros)
        return dict(sorted(contador.items()))


serie = input("Ingrese la serie de números separados por comas: ").split(",")

est = estadistica()

print("\nResultados Estadísticos:")
print(f"Media: {est.media(serie):.4f}")
print(f"Mediana: {est.mediana(serie):.4f}")
print(f"Moda: {est.moda(serie)}")
print(f"Varianza: {est.varianza(serie):.4f}")
print(f"Desviación Estándar: {est.desviacion(serie):.4f}")
print(f"Desviación Típica: {est.desviacion(serie):.4f}")
print(f"Rango: {est.rango(serie):.4f}")
print(f"Frecuencia: {est.frecuencia(serie)}")
```
